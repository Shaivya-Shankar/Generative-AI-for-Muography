# Dataset

This notebook uses histogram sampling to sample each of $(-\Delta P,\Delta \theta,\Delta \phi,\Delta r)$ individually based on their histograms.
To do so, from the histogram a PDF is obtained, and then its CDF is used through inverse sampling (see [here](https://en.wikipedia.org/wiki/Inverse_transform_sampling) for details).

In the arguments `P_bins` is used to seperate the sampling and histograming in different bins of momentum.

We recommend you first run the notebook with `P_bins` commented out, to sample marginally on P.
Then run agin with that argument, and see how much it improves things.

```python
## Get files #
data_path = '../data/'
material = 'Pb'
files = sorted(glob.glob(os.path.join(data_path,f'*{material}*.root')))
print (len(files))

# Main args #
args = {
    # Boilerplate arguments #
    'position_vars' : ['pos_x','pos_y','pos_z'],
    'position_unit' : units.cm,
    'momentum_vars' : ['p_x','p_y','p_z'],
    'momentum_unit' : units.MeV,
    'initial_energy_var' : 'initial_energy',
    'initial_energy_unit' : units.GeV,
    'dtype' : torch.float32,
    # Specific arguments #
    'bins' : [
        np.logspace(np.log10(1e-9),np.log10(200),100), # DP
        np.logspace(np.log10(1e-9),np.log10(1e-1),100),# Dtheta
        np.linspace(-math.pi,math.pi,100),             # Dphi
        np.linspace(1e-7,0.0563,100),                  # Dr
    ], # Bins for the different variables 
    'interpolate_sampling' : True, # Improve the CDF sampling
    'P_bins' : np.r_[0:10:1,10:100:10,100:1000:100], # fine binning
}

dataset_train = HistDataset(
    files = files[:20],
    make_sampler = True,
    **args
)
dataset_test = HistDataset(
    files = files[-1],
    make_sampler = False,
    mask_per_track = True,
    **args
)
```

```python
# We plot hre the PDF and CDF (potetially depending on te momentum, if you enabled it)
fig = dataset_train.plot_distribution(xscale='log',yscale='log',show=True)
```

```python
# To check the sampling, you can use this method
fig = dataset_train.plot_sampling(
    N = int(1e7),        # Number of samples to draw
    bins = 100,          # Number of bins for plotting the samples (can be different from the PDF bins as specified in the dataset)
    xscale = 'log',      # log or linear
    yscale = 'log',      # log or linear
    batch_size = None,   # if you ask for too large N, it will break memory, so you can use this to split the sampling into batches
    show = True,
)
```

# Propagation
