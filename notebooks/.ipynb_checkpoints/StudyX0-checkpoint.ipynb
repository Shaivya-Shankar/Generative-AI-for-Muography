```python
def plot_comparison(var,xlabel,logx=True,logy=True,normalise_x0=False):
    plt.figure()
    values = {}
    for material in sorted(X0_VALUES, key=lambda k: X0_VALUES[k]):
        x0 = X0_VALUES[material]
        dataset = datasets[material]
        val = ak.ravel(getattr(dataset,var)).to_numpy()
        if normalise_x0:
            val /= x0
        values[f'{material} [X0 = {x0:.2f}]'] = val
            
    all_values = np.concatenate([val for val in values.values()],axis=0)
    if logx:
        bins = np.logspace(math.log10(min(all_values)),math.log10(max(all_values)),101)
    else:
        bins = np.linspace(min(all_values),max(all_values),101)
    for label,val in values.items():
        plt.hist(
            val,
            bins = bins,
            label = label,
            histtype = 'step',
        )
    plt.legend(loc='center left', bbox_to_anchor=(1., 0.5))
    if logx:
        plt.xscale('log')
    if logy:
        plt.yscale('log')
    plt.xlabel(xlabel)
    plt.show()


```

```python
plot_comparison('dP',xlabel=r'$-\Delta P$',logx=True,logy=True)
plot_comparison('dtheta',xlabel=r'$\Delta \theta$',logx=True,logy=True)
plot_comparison('dphi',xlabel=r'$\Delta \phi$',logx=False,logy=True)
plot_comparison('dr',xlabel=r'$\Delta r$',logx=True,logy=True)
```

```python
plot_comparison('dP',xlabel=r'$-\frac{\Delta P}{X_0}$',logx=True,logy=True,normalise_x0=True)
plot_comparison('dtheta',xlabel=r'$\frac{\Delta \theta}{X_0}$',logx=True,logy=True,normalise_x0=True)
plot_comparison('dr',xlabel=r'$\frac{\Delta r}{X_0}$',logx=True,logy=True,normalise_x0=True)
```
