```python
%load_ext autoreload
%autoreload 2
%matplotlib inline

import os
import sys
from tqdm import tqdm
import math
import glob
import uproot
import torch
from torch import nn, optim
import vector
import numpy as np
import awkward as ak
import itertools
import matplotlib
import matplotlib.pyplot as plt
from hepunits import units

import lightning as L
from lightning.pytorch import loggers as pl_loggers

from torch.utils.data import DataLoader

vector.register_awkward()

src_path = os.path.abspath('../src/')
print (src_path)
if src_path not in sys.path:
    sys.path.append(src_path)

from datasets import *
from models import *
from callbacks import *
from annealing import *
from constants import X0_VALUES

print (f"Running on GPU : {torch.cuda.is_available()}")
accelerator = 'gpu' if torch.cuda.is_available() else 'cpu'
print (f"Accelerator : {accelerator}")
torch.set_float32_matmul_precision('medium')

X0_VALUES
```

```python
## Get files #
data_path = '../data/'
material = 'Pb'
files = sorted(glob.glob(os.path.join(data_path,f'*{material}*.root')))
print (len(files))

# Main args #
args = {
    # Boilerplate arguments #
    'position_vars' : ['pos_x','pos_y','pos_z'],
    'position_unit' : units.cm,
    'momentum_vars' : ['p_x','p_y','p_z'],
    'momentum_unit' : units.MeV,
    'initial_energy_var' : 'initial_energy',
    'initial_energy_unit' : units.GeV,
    'dtype' : torch.float32,
    'X0' : X0_VALUES[material]
}

dataset_train = VAEDataset(
    files = files[:1],
    **args
)
dataset_valid = VAEDataset(
    files = files[-2],
    scaling = dataset_train.scaling,
    **args
)
dataset_test = VAEDataset(
    files = files[-1],
    scaling = dataset_train.scaling,
    mask_per_track = True,
    **args
)


batch_size = 1024

loader_train = DataLoader(
    dataset = dataset_train,
    batch_size = batch_size,
    shuffle = True,
)
loader_valid = DataLoader(
    dataset = dataset_valid,
    batch_size = 10000,
    shuffle = False,
)
print (dataset_train.scaling)
```

```python
model = VariationalAutoencoder(
    dim_in = dataset_train.input_dim,
    dim_out = dataset_train.target_dim,
    dim_cond = dataset_train.condition_dim,
    dim_embed = 2,
    dim_latent = dataset_train.target_dim + dataset_train.condition_dim,
    variational = True,
    loss_function = nn.MSELoss(reduction='none'),
    neurons_embed   = [32]*2, 
    neurons_encoder = [64]*4,
    neurons_decoder = [64]*4,
    hidden_activation = nn.GELU,
    output_activation = None,
    batch_norm = True,
    free_bits = 0.,
)

batch = next(iter(loader_train))

losses = model.shared_eval(batch,0,'test')
print ('losses',losses)

samples = model.sample(batch,N=100)
print ('samples',samples.shape)

print (model)
```

```python
##### Parameters #####
epochs = 200
steps_per_epoch_train = math.ceil(len(dataset_train)/loader_train.batch_size)
print (f'Training   : Batch size = {loader_train.batch_size} => {steps_per_epoch_train} steps per epoch')

##### Optimizer #####
optimizer = optim.RAdam(model.parameters(), lr=1e-3)
model.set_optimizer(optimizer)

##### Annealing #####
kl_annealing = KLAnnealingCallback(
    annealing = CosineAnnealing(
        T = 50,
        beta_min = 1e-9,
        beta_max = 1e-3,
        warm_up = 0,
    ),
)

##### Scheduler #####
warmup_scheduler = torch.optim.lr_scheduler.ConstantLR(optimizer, factor = 1.0, total_iters = 50)
main_scheduler = torch.optim.lr_scheduler.ExponentialLR(optimizer, gamma = 0.95)
scheduler = torch.optim.lr_scheduler.SequentialLR(
    optimizer, 
    schedulers = [
        warmup_scheduler,
        main_scheduler,
    ], 
    milestones = [50],
)
model.set_scheduler_config(
    {
        'scheduler' : scheduler,
        'interval' : 'step' if isinstance(scheduler,optim.lr_scheduler.OneCycleLR) else 'epoch',
        'frequency' : 1,
        'monitor' : 'val/loss_tot',
        'strict' : True,
        'name' : 'scheduler',
    }
)

##### Callbacks #####
Ps = torch.logspace(math.log10(0.5),math.log10(500),51)
X0s = torch.tensor([X0_VALUES[material]])

sampling_callback = VAESamplingCallback(
    values = [Ps,X0s],
    scaling = dataset_train.scaling,
    plotting_config = dataset_train.plotting_config,
    preprocessed = False,
    frequency = 10,
    bins = 51,
    N_sample = 1000,
)

target_callback = VAEPredictionCallback(
    dataset = dataset_test,
    preprocessed = False,
    frequency = 10,
    bins = 51,
    batch_size = 10000,
)

callbacks = [
    L.pytorch.callbacks.LearningRateMonitor(logging_interval = 'epoch'),
    L.pytorch.callbacks.ModelSummary(max_depth=3),
    sampling_callback,
    target_callback,
    kl_annealing,
]

##### Logger #####
logger = pl_loggers.CometLogger(
    save_dir = '../comet_logs',
    project_name = 'muon-gen',
    experiment_name = 'VAE',
    offline = False,
)
logger.log_graph(model)

##### Trainer #####
trainer = L.Trainer(
    min_epochs = 1,
    max_epochs = epochs,
    callbacks = callbacks,
    devices = 'auto',
    accelerator = accelerator,
    logger = logger,
    # limit_train_batches = 5000, # can be used to make training faster (especially useful in dev)
    # limit_val_batches = 100,
)
##### Fit #####
trainer.fit(
    model = model,
    train_dataloaders = loader_train,
    val_dataloaders = loader_valid,
)
```
