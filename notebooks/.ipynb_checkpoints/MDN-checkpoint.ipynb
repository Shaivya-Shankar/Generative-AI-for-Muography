```python
%load_ext autoreload
%autoreload 2
%matplotlib inline

import os
import sys
from tqdm import tqdm
import math
import glob
import uproot
import torch
from torch import nn, optim
import vector
import numpy as np
import awkward as ak
import itertools
import matplotlib
import matplotlib.pyplot as plt
from hepunits import units

import lightning as L
from lightning.pytorch import loggers as pl_loggers

from torch.utils.data import DataLoader

vector.register_awkward()

src_path = os.path.abspath('../src/')
print (src_path)
if src_path not in sys.path:
    sys.path.append(src_path)

from datasets import *
from models import *
from callbacks import *
from annealing import *
from constants import X0_VALUES

print (f"Running on GPU : {torch.cuda.is_available()}")
accelerator = 'gpu' if torch.cuda.is_available() else 'cpu'
print (f"Accelerator : {accelerator}")
torch.set_float32_matmul_precision('medium')
```

```python
# Get files #
materials = list(X0_VALUES.keys())
# materials = ['Pb','U','Fe','Ca']
materials = ['Pb']

args = {
    'position_vars' : ['pos_x','pos_y','pos_z'],
    'position_unit' : units.cm,
    'momentum_vars' : ['p_x','p_y','p_z'],
    'momentum_unit' : units.MeV,
    'initial_energy_var' : 'initial_energy',
    'initial_energy_unit' : units.GeV,
    'dtype' : torch.float32,
}

datasets_train = {}
datasets_valid = {}
datasets_test = {}
for material in materials:
    files = sorted(glob.glob(f'../data/*{material}*root'))
    print ('='*60)
    print (material,materials)
    print ('Training dataset')
    datasets_train[material] = MDNDataset(
        files = files[:1],
        X0 = X0_VALUES[material],
        **args
    )
    print ('Validation dataset')
    datasets_valid[material] = MDNDataset(
        files = files[-2],
        X0 = X0_VALUES[material],
        scaling = datasets_train[material].scaling,
        **args
    )
for material in materials:
# for material in list(X0_VALUES.keys()):
    print ('Testing dataset')
    datasets_test[material] = MDNDataset(
        files = files[-1],
        X0 = X0_VALUES[material],
        **args
    )

print ('Common training dataset')
dataset_train = CombinedDataset(datasets_train)
print ('Common validation dataset')
dataset_valid = CombinedDataset(datasets_valid,scaling=dataset_train.scaling)
print ('Common test dataset')
dataset_test = CombinedDataset(datasets_test,scaling=dataset_train.scaling)

batch_size = 1024

loader_train = DataLoader(
    dataset = dataset_train,
    batch_size = batch_size,
    shuffle = True,
)
loader_valid = DataLoader(
    dataset = dataset_valid,
    batch_size = 10000,
    shuffle = False,
)

print (datasets_train)
print (dataset_train.scaling)
```

```python
model = MixtureDensityNetwork(
    backbone = DeepNeuralNetwork(
        dim_in = dataset_train.input_dim,
        dim_out = 64,
        hidden_activation = nn.GELU,
        output_activation = nn.GELU,
        neurons = [64]*5,
        batch_norm = True,
        # dropout = 0.,
    ),
    neurons = [],
    dim_out = dataset_train.target_dim,
    n_components = 3,
)


batch = next(iter(loader_train))
x = batch['input']
t = batch['target']
print (x.shape,t.shape)

y = model(batch)
pi,mu,sigma = y
print ('pi',pi.shape,'mu',mu.shape,'sigma',sigma.shape)

# loss_gaussian = model.log_likelihood_gaussian(y,t)
# loss_dist = model.log_likelihood(y,t)
# print (loss_gaussian.shape,loss_dist.shape)
loss = model.loss(y,t)
print ('loss',loss)
# print (loss['loss_tot'].shape)

samples = model.sample(batch,N=100)
print ('samples',samples.shape)

print (model)
```

```python
##### Parameters #####
epochs = 100
steps_per_epoch_train = math.ceil(len(dataset_train)/loader_train.batch_size)
print (f'Training   : Batch size = {loader_train.batch_size} => {steps_per_epoch_train} steps per epoch')

##### Optimizer #####
optimizer = optim.RAdam(model.parameters(), lr=1e-4)
model.set_optimizer(optimizer)

##### Scheduler #####
scheduler = torch.optim.lr_scheduler.ExponentialLR(optimizer, gamma = 0.9)
model.set_scheduler_config(
    {
        'scheduler' : scheduler,
        'interval' : 'step' if isinstance(scheduler,optim.lr_scheduler.OneCycleLR) else 'epoch',
        'frequency' : 1,
        'monitor' : 'val/loss_tot',
        'strict' : True,
        'name' : 'scheduler',
    }
)

##### Callbacks #####
# history_config = {
#     "train/loss_tot": {"label": "Training loss", "color": "r", "linestyle": "-"},
#     "val/loss_tot": {"label": "Validation loss", "color": "g", "linestyle": "--"},
#     "scheduler": {"label": "Learning rate"},
# }
# history_group = [("train/loss_tot", "val/loss_tot"), ("scheduler")]
# axes_params = [{}, {}]
# for metric in MDN.metrics.keys():
#     history_config[f'train/{metric}'] = {"label": f"Training {metric}", "color": "r", "linestyle": "-"}
#     history_config[f'val/{metric}'] = {"label": f"Validation {metric}", "color": "g", "linestyle": "--"}
#     history_group.append((f"train/{metric}", f"val/{metric}"))
#     axes_params.append({})
# history = HistoryCallback(
#     config = history_config,
#     groups =history_group,
#     axes_params = axes_params,
# )

callbacks = [
    # history,
    L.pytorch.callbacks.LearningRateMonitor(logging_interval = 'epoch'),
    L.pytorch.callbacks.ModelSummary(max_depth=2),
]

##### Logger #####
logger = pl_loggers.CometLogger(
    save_dir = '../comet_logs',
    project_name = 'muon-gen',
    experiment_name = 'MDN',
    offline = False,
)
logger.log_graph(model)

##### Trainer #####
trainer = L.Trainer(
    min_epochs = 1,
    max_epochs = epochs,
    callbacks = callbacks,
    devices = 'auto',
    accelerator = accelerator,
    logger = logger,
    log_every_n_steps = 100,
    limit_val_batches = 50,
)
##### Fit #####
trainer.fit(
    model = model,
    train_dataloaders = loader_train,
    val_dataloaders = loader_valid,
)
# history.plot()
```

```python
for idx in range(10):
    print ('scaled')
    fig,axs = plt.subplots(1,N_features,figsize=(5*N_features,5))
    if not isinstance(axs,np.ndarray):
        axs = [axs]
    for i in range(N_features):
        axs[i].hist(y[idx,i,:],bins=51,density=True,histtype='step',color="b",label='Pred (specific muon)')
        axs[i].axvline(t[idx,i],color='r',label='True (specific muon)')
        axs[i].hist(t[:,i],bins=51,density=True,histtype='step',color="g",label='True (all muons)')
        axs[i].legend()
        axs[i].set_yscale("log")
        axs[i].set_ylim(1e-1,None)
    plt.show()
    
    print ('raw')
    fig,axs = plt.subplots(1,N_features,figsize=(5*N_features,5))
    if not isinstance(axs,np.ndarray):
        axs = [axs]
    for i in range(N_features):
        axs[i].hist(y_raw[idx,i,:],bins=51,density=True,histtype='step',color="b",label='Pred (specific muon)')
        axs[i].axvline(t_raw[idx,i],color='r',label='True (specific muon)')
        axs[i].hist(t_raw[:,i],bins=51,density=True,histtype='step',color="g",label='True (all muons)')
        axs[i].legend()
        axs[i].set_ylim(1e-1,None)
        axs[i].set_yscale("log")
    plt.show()
    
```

```python
input_grid = torch.logspace(math.log10(0.5,),math.log10(500),101)
input_grid_scaled = dataset_train._scaling.transform(
    name = 'input',
    x = input_grid.unsqueeze(1),
    features = dataset_train._features['input']
)
with torch.no_grad():
    y_grid = MDN(input_grid_scaled,None)
pi,mu,sigma = y_grid

fig,axs = plt.subplots(nrows=MDN.dim_out,ncols=3,figsize=(15,9))
plt.subplots_adjust(wspace=0.2,hspace=0.2)
colors = plt.cm.rainbow(np.linspace(0, 1, MDN.n_components))
for j in range(MDN.dim_out):
    for i in range(MDN.n_components):
        axs[j,0].plot(input_grid,pi[:,i],color=colors[i])
        axs[j,1].plot(input_grid,mu[:,i,j],color=colors[i])
        axs[j,2].plot(input_grid,sigma[:,i,j],color=colors[i])
    axs[j,0].set_xscale('log')
    axs[j,1].set_xscale('log')
    axs[j,2].set_xscale('log')
    axs[j,0].set_xlabel(r'P')
    axs[j,1].set_xlabel(r'P')
    axs[j,2].set_xlabel(r'P')
    axs[j,0].set_ylabel(r'$\pi$')
    axs[j,1].set_ylabel(r'$\mu$')
    axs[j,2].set_ylabel(r'$\sigma$')

plt.show()
```

```python
input_grid_centers = (input_grid[1:]+input_grid[:-1])/2
input_grid_centers_scaled = dataset_train._scaling.transform(
    name = 'input',
    x = input_grid_centers.unsqueeze(-1),
    features = dataset_train._features['input']
)
N_sample = 10000
with torch.no_grad():
    y_grid = MDN.sample(input_grid_centers_scaled,None,N=N_sample)
y_grid = dataset_train._scaling.inverse(
    name = 'target',
    x = y_grid.permute(0,2,1).reshape(len(input_grid_centers)*N_sample,N_features),
    features = dataset_train._features['target'],
).reshape(len(input_grid_centers),N_sample,N_features).permute(0,2,1)
x_pos = torch.repeat_interleave(input_grid_centers.unsqueeze(-1),repeats=N_sample,dim=1)

fig,axs = plt.subplots(ncols=3,figsize=(18,5))
plt.subplots_adjust(wspace=0.3)
for j in range(MDN.dim_out):
    y_var = y_grid[:,j,:] 
    bins_y = torch.linspace(y_var.min(),y_var.max(),len(input_grid))
    h = axs[j].hist2d(x_pos.ravel(),y_var.ravel(),bins=(input_grid,bins_y),norm=matplotlib.colors.LogNorm())
    fig.colorbar(h[3],ax=axs[j])
    axs[j].set_xscale('log')
plt.show()    

fig = dataset_train.plot_variable_2D(name_x='input',name_y='target',raw=True,logx=True,logz=True)
```
