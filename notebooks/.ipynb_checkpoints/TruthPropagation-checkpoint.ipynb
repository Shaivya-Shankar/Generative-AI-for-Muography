# Propagation

The `Geant-like` implementation, here is the step-by-step propagation setup.

1. Initialise tracks with $\vec{r} = (0,0,0)$ and $\vec{P} = (0,0,P)$
2. Repeat the following for each N steps required by the user
- Generate $(-\Delta P,\Delta \theta,\Delta \phi,\Delta r)$
- Update the momentum $\vec{P}_{i+1} = \vec{P}_{i} + \Delta P $, using $(-\Delta P,\Delta \theta,\Delta \phi)$
- Obtain the muon direction $\hat{n} = \vec{P}_{i+1} / |\vec{P}_{i+1}|$
- Compute step $\Delta\vec{r} = \hat{n} \Delta r$
- Update the position $\vec{r}_{i+1} = \vec{r}_{i} + \Delta\vec{r}$

For a first check, let us use the true values of $(-\Delta P,\Delta \theta,\Delta \phi,\Delta r)$ to test the algorithm

```python
from propagation import TruthPropagator

# Initialise the propagator #
propagator = TruthPropagator(dataset=dataset,steps=1000)
```

```python
# Let us do a propagation once to see what it looks like #
# Idx are the indices of the muons you want to see
propagator.plot_sampling(idx=np.arange(10),N_sample=1)
```

```python
# You can also check the values relatively to the true=Geant
propagator.plot_sampling(idx=np.arange(10),N_sample=1,relative=True)
```

The relative errors above are probably due to truncation errors.

This should probably be solved in the future, or not ...

Now to make more checks, we will keep the same propagation using true $(-\Delta P,\Delta \theta,\Delta r)$ but sampling $\Delta \phi$ using a uniform distribution.
In practice, this means the propagation will now be stochastic, but still should be physically correct.

```python
propagator = TruthPropagator(dataset=dataset,sample_phi=True,steps=1000)
```

```python
# Let us propagate 10 times 
propagator.plot_sampling(idx=np.arange(10),N_sample=10)
```

```python
# Same, but you can check the relative values
propagator.plot_sampling(idx=np.arange(10),N_sample=10,relative=True)
```

Now we can continue generating more and more tracks to compare with the true/geant values, but it becomes hard to compare.
Instead, what we can do is generate a bunch of them, and plot the quantile coverages, i.e. the shape that contains eg 68% of the sampled tracks.

Note that for this to be accurate enough you need to generate many samples, it might take a while.

In the future, we will try to parallelize that.

```python
# Here we plot the 1/2 sigma, 1 sigma, and 2 sigma shapes
propagator.plot_sampling(idx=np.arange(5),N_sample=100,quantiles=[0.38,0.68,0.95],relative=False)
```

This above looks nice, but at the end of the day we want an objective measure of how good our model is.

However, similar to our sampling/generation, Geant4 is itself a stochastic process. So what we call "true" tracks could be done several times in Geant4 and return different tracks.

So to compare Geant4 with our generative model, which are both stochastic, we need a probability measure.
Below is the step by step thought process behind the plots below.
- First we take M true/Geant tracks
- For each of them we generate N sampled tracks
- We decide on a specific step to look at, eg 500
- We also decide on a specific quantile to check, eg 50%
- We will now count, at step 500, how many of the true tracks are within their 50% quantile shape as determined by the N sampled tracks

Now repeat that for a certain number of steps, and a certain number of quantiles, and you get the plot below.

An optimal generator would have a perfectly diagonal shape for each variable.

Note : the `batch_size` argument is to split the M true tracks into a smaller number to avoid blowing up the memory

... this might also be quite long

```python
propagator.plot_coverage(
    batch_size = 10000, # How many true tracks at a time
    N_sample   = 1000,   # How many generated/sampled tracks per true track
    steps      = [100,500,1000], # At which steps to make plot 
    quantiles  = np.r_[0:1:21j], # Quantiles to check in [0,1] range
)
```
